```python
from lsm import LSM

lsm = LSM("lsm_test")
lsm.open()
```

```python
lsm[b"i:1"] = b"1"
lsm[b"i:0"] = b"0"
lsm[b"q"] = b"_3"
lsm[b"i:2"] = b"2"
lsm[b"2"] = b"2"
```

```python
for k, v in lsm[b"i:":]:
    if not k.startswith(b"i:"):
        break
    print(k, v)
```

```python
for k, v in lsm[b"i:":]:
    if not k.startswith(b"i:"):
        break
    del lsm[k]
```

```python
lsm.delete(b"sdsd")
```

```python
with lsm.cursor() as cursor:
    while True:
        print(cursor.key(), cursor.value())
        if not cursor.next():
            break
```
